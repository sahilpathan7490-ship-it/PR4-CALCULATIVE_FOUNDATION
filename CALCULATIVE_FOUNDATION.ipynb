## 4.2 TRANSPODE AND INVERSE ( IF POSSIBLE)

```python
subjects = ["MATH", "SCIENCE", "ENGLISH", "HISTORY"]

# Students × Subjects Matrix
A = df[subjects].to_numpy()

# Transpose
A_T = A.T
print("Original:", A.shape)
print("Transpose:", A_T.shape)

# Inverse possible? Original 200×4 hai, isliye NO
print("Inverse of original matrix: Not possible (200×4 is not square)")

# 4×4 matrix se inverse
A_square = A[:4, :4]
det = np.linalg.det(A_square)

print("4×4 Determinant:", round(det, 2))

if not np.isclose(det, 0):
    A_inv = np.linalg.inv(A_square)
    print("Inverse:")
    print(np.round(A_inv, 4))
```

 5 . dataset has 4 subject dimensions:

MATH, SCIENCE, ENGLISH, HISTORY

So each student is represented as a 4D vector:

$$ Student = [Math,\ Science,\ English,\ History] $$

Example:

$$ S001=[73,66,61,62] $$
1. Line

A line is 1-dimensional.

In your dataset, if we vary only one direction, for example overall performance scaling:

$$ T(x)=x[73,66,61,62] $$

This represents a line through the origin in the 4D student-performance space.

2. Plane

A plane is 2-dimensional.

Take two students:

$$ S001=[73,66,61,62] $$ $$ S002=[71,83,95,79] $$

Their linear combinations

$$ aS001+bS002 $$

form a 2D plane inside the 4D dataset space.

3. Hyperplane

A hyperplane in 4D is 3-dimensional.

Using three independent student vectors:

$$ aS001+bS002+cS003 $$

creates a 3D hyperplane/subspace inside the 4D subject space.


conclusion: dataset has 4 dimensions, so students exist in a 4D feature space. Line = 1D, Plane = 2D, Hyperplane = 3D.

# 6  Dimensionality: 2D → 3D → Higher Dimensions

Hyperplane concept
2D: A line separates/represents points.
3D: A plane separates/represents points.
4D: A 3D hyperplane is used.
n dimensions: Hyperplane has n − 1 dimensio

```python
subjects = ["MATH", "SCIENCE", "ENGLISH", "HISTORY"]

X_2D = df[["MATH","SCIENCE"]]
X_3D = df[["MATH","SCIENCE","ENGLISH"]]
X_4D = df[subjects]

print("2D:", X_2D.shape)
print("3D:", X_3D.shape)
print("4D:", X_4D.shape)
print("4D Hyperplane dimension =", X_4D.shape[1] - 1)
```

# 7. Eigenvalues & Decomposition
# Compute the eigenvalues and eigenvectors of the covariance matrix.
1. Eigenvalue kya hota hai?

Eigenvalue batata hai ki covariance matrix mein kis direction mein data ka variance/spread kitna hai.

Eigenvector batata hai ki woh direction kya hai.

2. Covariance Matrix

Using all 200 students:

$$ C= \begin{bmatrix} 122.14&71.22&58.97&65.60\\ 71.22&123.36&62.97&62.68\\ 58.97&62.97&86.28&56.92\\ 65.60&62.68&56.92&99.71 \end{bmatrix} $$
3. Eigenvalues
$$ \boxed{ 299.149,\quad52.325,\quad45.722,\quad34.292 } $$
4. Eigenvectors

Corresponding eigenvectors:

[-0.539, -0.543, -0.436, -0.473]

[-0.644,  0.729,  0.110, -0.205]

[ 0.542,  0.335, -0.338, -0.692]

[-0.023,  0.247, -0.826,  0.506]

```python
X = df[["MATH","SCIENCE","ENGLISH","HISTORY"]].dropna().to_numpy()

C = np.cov(X, rowvar=False)

eigenvalues, eigenvectors = np.linalg.eigh(C)

idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("Covariance Matrix:\n", np.round(C,2))
print("\nEigenvalues:", np.round(eigenvalues,3))
print("\nEigenvectors:\n", np.round(eigenvectors,3))
```
